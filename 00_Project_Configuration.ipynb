```python
# ============================================================
# 00.2 PACKAGE INSTALLATION
# ============================================================

import subprocess

REQUIRED_PACKAGES = [
    "numpy",
    "pandas",
    "scipy",
    "scikit-learn",
    "pyyaml",
    "openpyxl",
    "matplotlib",
    "seaborn",
    "statsmodels",
]

def install_missing_packages(packages):
    for package in packages:
        import_name = package.replace("-", "_")
        try:
            __import__(import_name)
        except ImportError:
            subprocess.check_call([
                sys.executable,
                "-m",
                "pip",
                "install",
                "-q",
                package
            ])

install_missing_packages(REQUIRED_PACKAGES)

print("Required packages verified.")
```

```python
# ============================================================
# 00.7 DATASET DISCOVERY AND REGISTRY
# ============================================================

from pathlib import Path

# ------------------------------------------------------------
# SEARCH ROOT
# ------------------------------------------------------------

SEARCH_ROOT = Path("/content/drive/MyDrive")

# ------------------------------------------------------------
# EXPECTED DATASET IDENTIFIERS
# ------------------------------------------------------------

DATASET_IDENTIFIERS = {
    "adult_income": [
        "adult_income.csv",
        "adult_income_processed.csv",
    ],

    "bank_marketing": [
        "bank_marketing.csv",
        "bank_marketing_processed.csv",
    ],

    "diabetes_130us": [
        "diabetes_130us.csv",
        "diabetes_130us_processed.csv",
    ],
}

# ------------------------------------------------------------
# DISCOVER EXISTING FILES
# ------------------------------------------------------------

DISCOVERED_DATASETS = {}

for dataset_id, filenames in DATASET_IDENTIFIERS.items():

    matches = []

    for filename in filenames:
        matches.extend(
            SEARCH_ROOT.rglob(filename)
        )

    # Remove duplicate paths
    matches = sorted(
        set(matches),
        key=lambda p: str(p)
    )

    DISCOVERED_DATASETS[dataset_id] = matches

# ------------------------------------------------------------
# DISPLAY DISCOVERY RESULTS
# ------------------------------------------------------------

print("=" * 90)
print("DATASET DISCOVERY")
print("=" * 90)

for dataset_id, matches in DISCOVERED_DATASETS.items():

    print(f"\n{dataset_id}")

    if not matches:
        print("  NOT FOUND")

    else:
        for path in matches:
            print(f"  {path}")
```

```python
# ============================================================
# 00.14 SAVE FINAL CONFIGURATION
# ============================================================

def make_yaml_safe(obj):

    if isinstance(obj, dict):
        return {
            str(key): make_yaml_safe(value)
            for key, value in obj.items()
        }

    if isinstance(obj, (list, tuple)):
        return [
            make_yaml_safe(value)
            for value in obj
        ]

    if isinstance(obj, Path):
        return str(obj)

    if isinstance(obj, np.integer):
        return int(obj)

    if isinstance(obj, np.floating):
        return float(obj)

    if isinstance(obj, np.bool_):
        return bool(obj)

    if obj is None:
        return None

    if isinstance(obj, (str, int, float, bool)):
        return obj

    return str(obj)


FINAL_CONFIG = {

    "project": {
        "name": "AIR-LLM",
        "version": "1.0",
        "purpose": (
            "Adaptive LLM-guided missing-value "
            "imputation research framework"
        ),
        "created_at": str(
            datetime.now().isoformat(
                timespec="seconds"
            )
        ),
    },

    "project_root": str(PROJECT_ROOT),

    "datasets": DATASET_REGISTRY,

    "experiment": EXPERIMENT_CONFIG,

    "missingness": MISSINGNESS_CONFIG,

    "candidate_strategies": CANDIDATE_STRATEGIES,

    "evaluation": EVALUATION_CONFIG,

    "llm": LLM_CONFIG,

    "reproducibility": REPRODUCIBILITY_CONFIG,

    "device": DEVICE_CONFIG,
}

# ------------------------------------------------------------
# Convert everything to YAML-compatible native Python types
# ------------------------------------------------------------

FINAL_CONFIG = make_yaml_safe(FINAL_CONFIG)

# ------------------------------------------------------------
# Configuration paths
# ------------------------------------------------------------

CONFIG_PATH = (
    DIRECTORIES["config"] /
    "air_llm_config.yaml"
)

DATASET_REGISTRY_PATH = (
    DIRECTORIES["config"] /
    "dataset_registry.yaml"
)

EXPERIMENT_REGISTRY_PATH = (
    DIRECTORIES["config"] /
    "experiment_registry.yaml"
)

# ------------------------------------------------------------
# Save master configuration
# ------------------------------------------------------------

with open(
    CONFIG_PATH,
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        FINAL_CONFIG,
        f,
        sort_keys=False,
        allow_unicode=True
    )

# ------------------------------------------------------------
# Save dataset registry
# ------------------------------------------------------------

DATASET_REGISTRY_SAFE = make_yaml_safe(
    DATASET_REGISTRY
)

with open(
    DATASET_REGISTRY_PATH,
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        DATASET_REGISTRY_SAFE,
        f,
        sort_keys=False,
        allow_unicode=True
    )

# ------------------------------------------------------------
# Experiment registry
# ------------------------------------------------------------

EXPERIMENT_REGISTRY = {

    "datasets": DATASET_IDS,

    "missingness_mechanisms":
        MISSINGNESS_MECHANISMS,

    "missingness_rates":
        MISSINGNESS_RATES,

    "candidate_strategies":
        CANDIDATE_STRATEGY_IDS,

    "evaluation_groups":
        list(EVALUATION_CONFIG.keys()),

    "n_repetitions":
        EXPERIMENT_CONFIG["n_repetitions"],

    "seed": SEED,
}

EXPERIMENT_REGISTRY = make_yaml_safe(
    EXPERIMENT_REGISTRY
)

with open(
    EXPERIMENT_REGISTRY_PATH,
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        EXPERIMENT_REGISTRY,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("=" * 90)
print("FINAL CONFIGURATION SAVED SUCCESSFULLY")
print("=" * 90)

print(
    f"Master config       : {CONFIG_PATH}"
)

print(
    f"Dataset registry    : {DATASET_REGISTRY_PATH}"
)

print(
    f"Experiment registry : {EXPERIMENT_REGISTRY_PATH}"
)
```

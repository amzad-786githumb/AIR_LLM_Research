```python
# ============================================================
# CELL 00.15.2 — FINAL MASTER YAML CONFIGURATION EXPORT
# ============================================================
#
# Purpose:
#   1. Convert every configuration value to YAML-safe
#      native Python objects.
#   2. Remove library-specific objects.
#   3. Save the permanent master configuration to Drive.
#   4. Reload the YAML immediately.
#   5. Validate the saved configuration.
#   6. Save a configuration manifest for reproducibility.
# ============================================================

from pathlib import Path
import json
import numbers
import yaml
import numpy as np


# ------------------------------------------------------------
# 1. UNIVERSAL YAML-SAFE CONVERSION
# ------------------------------------------------------------

def _yaml_safe(value):

    if value is None:
        return None

    if type(value) is str:
        return value

    if type(value) is int:
        return value

    if type(value) is float:
        return value

    if type(value) is bool:
        return value

    if isinstance(value, Path):
        return str(value)

    if isinstance(value, np.generic):
        return _yaml_safe(value.item())

    if isinstance(value, np.ndarray):
        return [
            _yaml_safe(v)
            for v in value.tolist()
        ]

    if isinstance(value, dict):
        return {
            str(_yaml_safe(key)): _yaml_safe(val)
            for key, val in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [
            _yaml_safe(v)
            for v in value
        ]

    if isinstance(value, set):
        return [
            _yaml_safe(v)
            for v in sorted(
                value,
                key=lambda x: str(x)
            )
        ]

    if isinstance(value, numbers.Integral):
        return int(value)

    if isinstance(value, numbers.Real):
        return float(value)

    return str(value)


# ------------------------------------------------------------
# 2. REQUIRED OBJECTS
# ------------------------------------------------------------

if "MASTER_CONFIG" not in globals():
    raise RuntimeError(
        "MASTER_CONFIG is not defined. "
        "Run CELL 00.15.1 first."
    )

if "DIRECTORIES" not in globals():
    raise RuntimeError(
        "DIRECTORIES is not defined. "
        "Run CELL 00.14.1 first."
    )


# ------------------------------------------------------------
# 3. CONVERT MASTER CONFIGURATION
# ------------------------------------------------------------

MASTER_CONFIG_SERIALIZABLE = _yaml_safe(
    MASTER_CONFIG
)


# ------------------------------------------------------------
# 4. SERIALIZATION SAFETY CHECK
# ------------------------------------------------------------

try:

    json.dumps(
        MASTER_CONFIG_SERIALIZABLE,
        ensure_ascii=False
    )

except TypeError as error:

    raise TypeError(
        "MASTER_CONFIG contains a non-serializable object "
        f"after conversion: {error}"
    )


# ------------------------------------------------------------
# 5. CONFIGURATION DIRECTORY
# ------------------------------------------------------------

CONFIG_DIR = Path(
    DIRECTORIES["config"]
)

CONFIG_DIR.mkdir(
    parents=True,
    exist_ok=True
)

if not os.access(CONFIG_DIR, os.W_OK):
    raise PermissionError(
        f"Configuration directory is not writable: {CONFIG_DIR}"
    )


# ------------------------------------------------------------
# 6. MASTER YAML PATH
# ------------------------------------------------------------

CONFIG_PATH = (
    CONFIG_DIR / "config.yaml"
)


# ------------------------------------------------------------
# 7. SAVE MASTER YAML
# ------------------------------------------------------------

with open(
    CONFIG_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        MASTER_CONFIG_SERIALIZABLE,
        file,
        sort_keys=False,
        allow_unicode=True,
        default_flow_style=False,
        indent=2,
        width=120
    )


# ------------------------------------------------------------
# 8. VERIFY SAVED FILE
# ------------------------------------------------------------

if not CONFIG_PATH.exists():
    raise RuntimeError(
        "Master configuration file was not created."
    )

if CONFIG_PATH.stat().st_size == 0:
    raise RuntimeError(
        "Master configuration file is empty."
    )


# ------------------------------------------------------------
# 9. RELOAD SAVED YAML
# ------------------------------------------------------------

with open(
    CONFIG_PATH,
    "r",
    encoding="utf-8"
) as file:

    MASTER_CONFIG_LOADED = yaml.safe_load(file)


if not isinstance(
    MASTER_CONFIG_LOADED,
    dict
):

    raise RuntimeError(
        "Reloaded MASTER_CONFIG is not a dictionary."
    )


# ------------------------------------------------------------
# 10. VERIFY PROJECT ROOT
# ------------------------------------------------------------

saved_project_root = str(
    MASTER_CONFIG_SERIALIZABLE["project_root"]
)

expected_project_root = str(
    PROJECT_ROOT
)

assert (
    saved_project_root == expected_project_root
), (
    "PROJECT_ROOT mismatch in saved configuration.\n"
    f"Expected: {expected_project_root}\n"
    f"Saved:    {saved_project_root}"
)


# ------------------------------------------------------------
# 11. REQUIRED CONFIGURATION SECTIONS
# ------------------------------------------------------------

EXPECTED_CONFIG_SECTIONS = [

    "project",
    "environment",
    "reproducibility",
    "project_root",
    "directories",
    "datasets",
    "dataset_paths",
    "feature_target",
    "feature_type_policy",
    "missingness",
    "split",
    "candidate_strategies",
    "strategy_selection_policy",
    "llm",
    "llm_output_schema",
    "evaluation",
    "statistical",
    "robustness",
    "utility_weights",
    "utility_score_definition",
    "confidence_thresholds",
    "confidence_policy",
    "experiment_scenarios",
    "experiment_design",
    "notebook_registry"
]

MISSING_SECTIONS = [
    section
    for section in EXPECTED_CONFIG_SECTIONS
    if section not in MASTER_CONFIG_SERIALIZABLE
]

if MISSING_SECTIONS:

    raise RuntimeError(
        "Master configuration is incomplete. "
        f"Missing sections: {MISSING_SECTIONS}"
    )


# ------------------------------------------------------------
# 12. VERIFY RELOADED CONFIGURATION
# ------------------------------------------------------------

for section in EXPECTED_CONFIG_SECTIONS:

    assert section in MASTER_CONFIG_LOADED, (
        f"Reloaded YAML is missing section: {section}"
    )


# ------------------------------------------------------------
# 13. VERIFY DATASET REGISTRY
# ------------------------------------------------------------

SAVED_DATASETS = (
    MASTER_CONFIG_LOADED["datasets"]
)

assert isinstance(
    SAVED_DATASETS,
    dict
)

assert set(SAVED_DATASETS.keys()) == set(
    DATASET_REGISTRY.keys()
)


# ------------------------------------------------------------
# 14. VERIFY DATASET PATH REGISTRY
# ------------------------------------------------------------

SAVED_DATASET_PATHS = (
    MASTER_CONFIG_LOADED["dataset_paths"]
)

assert set(
    SAVED_DATASET_PATHS.keys()
) == set(
    DATASET_PATHS.keys()
)


# ------------------------------------------------------------
# 15. WRITE CONFIGURATION MANIFEST
# ------------------------------------------------------------

MANIFEST_PATH = (
    CONFIG_DIR
    / "configuration_manifest.json"
)

MANIFEST = {

    "configuration_file":
        str(CONFIG_PATH),

    "configuration_size_bytes":
        int(CONFIG_PATH.stat().st_size),

    "configuration_sections":
        list(
            MASTER_CONFIG_SERIALIZABLE.keys()
        ),

    "expected_sections":
        EXPECTED_CONFIG_SECTIONS,

    "missing_required_sections":
        MISSING_SECTIONS,

    "dataset_count":
        len(SAVED_DATASETS),

    "datasets":
        list(SAVED_DATASETS.keys()),

    "yaml_serialization":
        "PASSED",

    "yaml_reload":
        "PASSED",

    "json_serialization":
        "PASSED",

    "project_root_validation":
        "PASSED",

    "dataset_registry_validation":
        "PASSED",

    "dataset_path_validation":
        "PASSED"
}

with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        MANIFEST,
        file,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 16. FINAL OUTPUT
# ------------------------------------------------------------

print()
print("=" * 100)
print("AIR-LLM MASTER CONFIGURATION EXPORT")
print("=" * 100)

print(
    f"Master YAML configuration : {CONFIG_PATH}"
)

print(
    f"Configuration size         : "
    f"{CONFIG_PATH.stat().st_size:,} bytes"
)

print(
    f"Manifest                   : {MANIFEST_PATH}"
)

print(
    "YAML serialization         : PASSED"
)

print(
    "YAML reload validation     : PASSED"
)

print(
    "JSON serialization         : PASSED"
)

print(
    "Project root validation    : PASSED"
)

print(
    "Dataset registry           : PASSED"
)

print(
    "Dataset path registry      : PASSED"
)

print(
    f"Configuration sections     : "
    f"{len(MASTER_CONFIG_SERIALIZABLE)}"
)

print()
print("Required configuration sections: ")

for section in EXPECTED_CONFIG_SECTIONS:

    print(f"  [OK] {section}")

print("=" * 100)
print("CELL 00.15.2 COMPLETED SUCCESSFULLY")
print("=" * 100)
```

```python
# ============================================================
# NOTEBOOK 09.1 — LOAD PREVIOUS NOTEBOOK ARTIFACTS
# ============================================================

print("=" * 100)
print("AIR-LLM — NOTEBOOK 09.1")
print("LOADING PREVIOUS NOTEBOOK ARTIFACTS")
print("=" * 100)


# ------------------------------------------------------------
# Helper: recursive CSV discovery
# ------------------------------------------------------------

def find_csv_files(root):

    if not root.exists():
        return []

    return sorted(
        root.rglob("*.csv")
    )


def find_json_files(root):

    if not root.exists():
        return []

    return sorted(
        root.rglob("*.json")
    )


# ------------------------------------------------------------
# Discover artifacts
# ------------------------------------------------------------

SEARCH_ROOTS = {
    "notebook_04": NOTEBOOK_04_DIR,
    "notebook_05": NOTEBOOK_05_DIR,
    "notebook_06": NOTEBOOK_06_DIR,
    "notebook_07": NOTEBOOK_07_DIR,
    "notebook_08": NOTEBOOK_08_DIR
}


ARTIFACT_FILES = {}

for notebook_name, root in SEARCH_ROOTS.items():

    ARTIFACT_FILES[
        notebook_name
    ] = (
        find_csv_files(root)
        +
        find_json_files(root)
    )


print("\nDISCOVERED ARTIFACTS")
print("-" * 100)

for notebook_name, files in ARTIFACT_FILES.items():

    print(
        f"{notebook_name:15s}: "
        f"{len(files)} files"
    )


print("\nArtifact discovery completed.")
```

```python
# ============================================================
# NOTEBOOK 09.2 — IDENTIFY DECISION ARTIFACTS
# ============================================================

print("=" * 100)
print("AIR-LLM — NOTEBOOK 09.2")
print("IDENTIFYING DECISION ARTIFACTS")
print("=" * 100)


def read_csv_safe(path):

    try:

        return pd.read_csv(path)

    except Exception as exc:

        print(
            f"WARNING: Could not read {path.name}: {exc}"
        )

        return None


# ------------------------------------------------------------
# Search keywords
# ------------------------------------------------------------

SELECTION_KEYWORDS = [
    "selection",
    "selected",
    "utility",
    "adaptive"
]

CONFIDENCE_KEYWORDS = [
    "confidence",
    "abstention",
    "decision"
]

RECOMMENDATION_KEYWORDS = [
    "recommendation",
    "recommend",
    "llm"
]


def classify_file(path):

    name = path.stem.lower()

    if any(
        key in name
        for key in SELECTION_KEYWORDS
    ):
        return "selection"

    if any(
        key in name
        for key in CONFIDENCE_KEYWORDS
    ):
        return "confidence"

    if any(
        key in name
        for key in RECOMMENDATION_KEYWORDS
    ):
        return "recommendation"

    return "other"


CLASSIFIED_FILES = {
    "selection": [],
    "confidence": [],
    "recommendation": [],
    "other": []
}


for notebook_name, files in ARTIFACT_FILES.items():

    for path in files:

        category = classify_file(path)

        CLASSIFIED_FILES[
            category
        ].append(path)


for category, files in CLASSIFIED_FILES.items():

    print(
        f"{category:18s}: "
        f"{len(files)}"
    )


print("\nCandidate decision artifacts identified.")
```

```python
# ============================================================
# NOTEBOOK 09.8 — DATA EVIDENCE EXTRACTION
# ============================================================

print("=" * 100)
print("AIR-LLM — NOTEBOOK 09.8")
print("EXTRACTING DATA EVIDENCE")
print("=" * 100)


def first_matching_column(df, keywords):

    if df is None:
        return None

    for column in df.columns:

        name = str(column).lower()

        for keyword in keywords:

            if keyword in name:
                return column

    return None


FEATURE_DATA = FEATURE_PROFILE_DF.copy()


FP_DATASET_COL = find_column(
    FEATURE_DATA,
    ["dataset_id", "dataset"]
)

FP_FEATURE_COL = find_column(
    FEATURE_DATA,
    ["feature", "feature_name", "column"]
)

FP_TYPE_COL = find_column(
    FEATURE_DATA,
    ["feature_type", "dtype", "type"]
)


def profile_lookup(dataset_id, feature):

    subset = FEATURE_DATA

    if FP_DATASET_COL is not None:

        subset = subset[
            subset[FP_DATASET_COL].astype(str)
            ==
            str(dataset_id)
        ]

    if FP_FEATURE_COL is not None:

        subset = subset[
            subset[FP_FEATURE_COL].astype(str)
            ==
            str(feature)
        ]

    if subset.empty:

        return {}

    return subset.iloc[0].to_dict()


def missingness_lookup(dataset_id, feature):

    subset = MISSINGNESS_PROFILE_DF

    dataset_col = find_column(
        subset,
        ["dataset_id", "dataset"]
    )

    feature_col = find_column(
        subset,
        ["feature", "feature_name", "column"]
    )

    if dataset_col is not None:

        subset = subset[
            subset[dataset_col].astype(str)
            ==
            str(dataset_id)
        ]

    if feature_col is not None:

        subset = subset[
            subset[feature_col].astype(str)
            ==
            str(feature)
        ]

    if subset.empty:

        return {}

    return subset.iloc[0].to_dict()


print(
    "Profile lookup functions created."
)
```

```python
# ============================================================
# NOTEBOOK 09.9 — EMPIRICAL EVIDENCE EXTRACTION
# ============================================================

print("=" * 100)
print("AIR-LLM — NOTEBOOK 09.9")
print("EXTRACTING EMPIRICAL EVIDENCE")
print("=" * 100)


# ------------------------------------------------------------
# Utility components
# ------------------------------------------------------------

UTILITY_COMPONENTS = [
    "accuracy",
    "distribution",
    "dependency",
    "predictive",
    "efficiency"
]


def extract_metric(row, keywords):

    for column in row.index:

        name = str(column).lower()

        if any(
            keyword in name
            for keyword in keywords
        ):

            value = safe_float(
                row[column]
            )

            if np.isfinite(value):

                return value

    return np.nan


def empirical_evidence(row):

    evidence = {}

    evidence["accuracy"] = extract_metric(
        row,
        [
            "accuracy",
            "mae",
            "rmse",
            "f1",
            "reconstruction"
        ]
    )

    evidence["distribution"] = extract_metric(
        row,
        [
            "distribution",
            "wasserstein",
            "js",
            "kl"
        ]
    )

    evidence["dependency"] = extract_metric(
        row,
        [
            "dependency",
            "correlation",
            "spearman",
            "mutual_information"
        ]
    )

    evidence["predictive"] = extract_metric(
        row,
        [
            "predictive",
            "downstream",
            "prediction"
        ]
    )

    evidence["efficiency"] = extract_metric(
        row,
        [
            "efficiency",
            "runtime",
            "time"
        ]
    )

    return evidence


print(
    "Empirical evidence extraction ready."
)
```

```python
# ============================================================
# NOTEBOOK 09.11 — LLM RECOMMENDATION EVIDENCE
# ============================================================

print("=" * 100)
print("AIR-LLM — NOTEBOOK 09.11")
print("MATCHING LLM RECOMMENDATION EVIDENCE")
print("=" * 100)


def recommendation_lookup(
    dataset_id,
    feature,
    strategy
):

    if RECOMMENDATION_DF is None:

        return {}

    df = RECOMMENDATION_DF.copy()

    dataset_col = find_column(
        df,
        ["dataset_id", "dataset"]
    )

    feature_col = find_column(
        df,
        ["feature", "feature_name", "column"]
    )

    strategy_col = find_column(
        df,
        [
            "strategy",
            "method",
            "candidate_strategy"
        ]
    )

    if dataset_col is not None:

        df = df[
            df[dataset_col].astype(str)
            ==
            str(dataset_id)
        ]

    if feature_col is not None:

        df = df[
            df[feature_col].astype(str)
            ==
            str(feature)
        ]

    if strategy_col is not None:

        df = df[
            df[strategy_col].astype(str).str.lower()
            ==
            str(strategy).lower()
        ]

    if df.empty:

        return {}

    return df.iloc[0].to_dict()


def extract_recommendation_text(data):

    if not data:

        return ""

    preferred = [
        "rationale",
        "reason",
        "explanation",
        "recommendation_reason"
    ]

    for preferred_key in preferred:

        for key, value in data.items():

            if str(key).lower() == preferred_key:

                text = safe_text(
                    value,
                    ""
                )

                if text:

                    return text

    return ""


def extract_llm_rank(data):

    if not data:

        return np.nan

    for key, value in data.items():

        if "rank" in str(key).lower():

            result = safe_float(value)

            if np.isfinite(result):

                return result

    return np.nan


print(
    "LLM recommendation evidence functions ready."
)
```

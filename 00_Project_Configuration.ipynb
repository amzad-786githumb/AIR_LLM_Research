```python
# ============================================================
# CELL 00.15.2 — FINAL MASTER YAML CONFIGURATION EXPORT
# ============================================================
#
# Purpose:
#   1. Convert EVERY configuration value to a YAML-safe
#      native Python object.
#   2. Remove library-specific objects such as torch version
#      objects, NumPy scalars, Path objects, etc.
#   3. Save the permanent master configuration.
#   4. Reload the YAML immediately.
#   5. Validate that the saved configuration is usable by
#      all future AIR-LLM research notebooks.
#
# This cell is intentionally self-contained.
# ============================================================

from pathlib import Path
import json
import numbers
import yaml
import numpy as np


# ------------------------------------------------------------
# 1. UNIVERSAL YAML-SAFE CONVERSION
# ------------------------------------------------------------

def _yaml_safe(value):
    """
    Recursively convert arbitrary Python/library objects into
    standard Python objects supported by yaml.safe_dump().
    """

    # None
    if value is None:
        return None

    # Native primitive types
    if type(value) is str:
        return value

    if type(value) is int:
        return value

    if type(value) is float:
        return value

    if type(value) is bool:
        return value

    # pathlib.Path
    if isinstance(value, Path):
        return str(value)

    # NumPy scalar types
    if isinstance(value, np.generic):
        return _yaml_safe(value.item())

    # NumPy arrays
    if isinstance(value, np.ndarray):
        return [
            _yaml_safe(v)
            for v in value.tolist()
        ]

    # Dictionaries
    if isinstance(value, dict):
        return {
            str(_yaml_safe(key)): _yaml_safe(val)
            for key, val in value.items()
        }

    # Lists / tuples
    if isinstance(value, (list, tuple)):
        return [
            _yaml_safe(v)
            for v in value
        ]

    # Sets
    if isinstance(value, set):
        return [
            _yaml_safe(v)
            for v in sorted(
                value,
                key=lambda x: str(x)
            )
        ]

    # Numeric library objects
    if isinstance(value, numbers.Integral):
        return int(value)

    if isinstance(value, numbers.Real):
        return float(value)

    # Objects such as torch.__version__,
    # packaging.version.Version, etc.
    #
    # Convert everything remaining into an ordinary string.
    return str(value)


# ------------------------------------------------------------
# 2. REBUILD MASTER CONFIGURATION
# ------------------------------------------------------------

if "MASTER_CONFIG" not in globals():
    raise RuntimeError(
        "MASTER_CONFIG is not defined. "
        "Run the previous configuration cells "
        "00.1–00.14 before running CELL 00.15.2."
    )

MASTER_CONFIG_SERIALIZABLE = _yaml_safe(
    MASTER_CONFIG
)


# ------------------------------------------------------------
# 3. FINAL SAFETY CHECK
# ------------------------------------------------------------

# Convert to JSON first.
#
# JSON only accepts standard Python types and therefore provides
# an additional guarantee that no library-specific object remains.

try:

    json.dumps(
        MASTER_CONFIG_SERIALIZABLE,
        ensure_ascii=False
    )

except TypeError as error:

    raise TypeError(
        "MASTER_CONFIG still contains a non-serializable "
        f"object after conversion: {error}"
    )


# ------------------------------------------------------------
# 4. ENSURE CONFIGURATION DIRECTORY EXISTS
# ------------------------------------------------------------

if "DIRECTORIES" not in globals():
    raise RuntimeError(
        "DIRECTORIES is not defined. "
        "Run CELL 00.14 before CELL 00.15.2."
    )

CONFIG_DIR = Path(
    DIRECTORIES["config"]
)

CONFIG_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 5. MASTER YAML PATH
# ------------------------------------------------------------

CONFIG_PATH = (
    CONFIG_DIR / "config.yaml"
)


# ------------------------------------------------------------
# 6. SAVE MASTER YAML
# ------------------------------------------------------------

with open(
    CONFIG_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        MASTER_CONFIG_SERIALIZABLE,
        file,
        sort_keys=False,
        allow_unicode=True,
        default_flow_style=False,
        indent=2,
        width=120
    )


# ------------------------------------------------------------
# 7. VERIFY FILE EXISTS
# ------------------------------------------------------------

if not CONFIG_PATH.exists():
    raise RuntimeError(
        "Master configuration file was not created."
    )

if CONFIG_PATH.stat().st_size == 0:
    raise RuntimeError(
        "Master configuration file is empty."
    )


# ------------------------------------------------------------
# 8. RELOAD SAVED YAML
# ------------------------------------------------------------

with open(
    CONFIG_PATH,
    "r",
    encoding="utf-8"
) as file:

    MASTER_CONFIG_LOADED = yaml.safe_load(file)


# ------------------------------------------------------------
# 9. VALIDATE RELOADED CONFIGURATION
# ------------------------------------------------------------

if not isinstance(
    MASTER_CONFIG_LOADED,
    dict
):

    raise RuntimeError(
        "Reloaded MASTER_CONFIG is not a dictionary."
    )


# ------------------------------------------------------------
# 10. VERIFY PROJECT ROOT
# ------------------------------------------------------------

if "project_root" in MASTER_CONFIG_SERIALIZABLE:

    saved_project_root = str(
        MASTER_CONFIG_SERIALIZABLE["project_root"]
    )

    if "PROJECT_ROOT" in globals():

        expected_project_root = str(
            PROJECT_ROOT
        )

        if saved_project_root != expected_project_root:

            raise RuntimeError(
                "PROJECT_ROOT mismatch in saved configuration.\n"
                f"Expected: {expected_project_root}\n"
                f"Saved:    {saved_project_root}"
            )


# ------------------------------------------------------------
# 11. VERIFY DATASET REGISTRY
# ------------------------------------------------------------

if "datasets" in MASTER_CONFIG_SERIALIZABLE:

    saved_datasets = (
        MASTER_CONFIG_SERIALIZABLE["datasets"]
    )

    if not isinstance(
        saved_datasets,
        dict
    ):

        raise RuntimeError(
            "Dataset registry is not a dictionary."
        )


# ------------------------------------------------------------
# 12. VERIFY IMPORTANT CONFIGURATION SECTIONS
# ------------------------------------------------------------

EXPECTED_CONFIG_SECTIONS = [

    "datasets",
    "features",
    "missingness",
    "candidate_strategies",
    "llm",
    "evaluation",
    "utility_weights",
    "confidence_thresholds",
    "experiment_scenarios"
]

PRESENT_SECTIONS = [
    key
    for key in EXPECTED_CONFIG_SECTIONS
    if key in MASTER_CONFIG_SERIALIZABLE
]

MISSING_SECTIONS = [
    key
    for key in EXPECTED_CONFIG_SECTIONS
    if key not in MASTER_CONFIG_SERIALIZABLE
]

if MISSING_SECTIONS:

    print(
        "WARNING — Configuration sections not found:"
    )

    for section in MISSING_SECTIONS:
        print(f"  - {section}")


# ------------------------------------------------------------
# 13. WRITE A HUMAN-READABLE CONFIGURATION MANIFEST
# ------------------------------------------------------------

MANIFEST_PATH = (
    CONFIG_DIR
    / "configuration_manifest.json"
)

MANIFEST = {

    "configuration_file":
        str(CONFIG_PATH),

    "configuration_size_bytes":
        int(CONFIG_PATH.stat().st_size),

    "configuration_sections":
        list(
            MASTER_CONFIG_SERIALIZABLE.keys()
        ),

    "expected_sections":
        EXPECTED_CONFIG_SECTIONS,

    "present_required_sections":
        PRESENT_SECTIONS,

    "missing_required_sections":
        MISSING_SECTIONS,

    "yaml_reload_success":
        True,

    "yaml_serialization":
        "PASSED",

    "json_serialization":
        "PASSED"
}


with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        MANIFEST,
        file,
        indent=2,
        ensure_ascii=False
    )


# ------------------------------------------------------------
# 14. FINAL OUTPUT
# ------------------------------------------------------------

print()
print("=" * 100)
print("AIR-LLM MASTER CONFIGURATION EXPORT")
print("=" * 100)

print(
    f"Master YAML configuration : {CONFIG_PATH}"
)

print(
    f"Configuration size         : "
    f"{CONFIG_PATH.stat().st_size:,} bytes"
)

print(
    f"Manifest                   : {MANIFEST_PATH}"
)

print(
    f"YAML serialization         : PASSED"
)

print(
    f"YAML reload validation     : PASSED"
)

print(
    f"JSON serialization         : PASSED"
)

print(
    f"Configuration sections     : "
    f"{len(MASTER_CONFIG_SERIALIZABLE)}"
)

print()
print("Required configuration sections:")

for section in EXPECTED_CONFIG_SECTIONS:

    status = (
        "OK"
        if section in MASTER_CONFIG_SERIALIZABLE
        else "MISSING"
    )

    print(
        f"  [{status}] {section}"
    )

print("=" * 100)
print("CELL 00.15.2 COMPLETED SUCCESSFULLY")
print("=" * 100)
```
